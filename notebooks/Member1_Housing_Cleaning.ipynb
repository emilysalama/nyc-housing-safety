```python
import pandas as pd
import numpy as np

housing_id = "w9ak-ipjd"

url = f"https://data.cityofnewyork.us/resource/{housing_id}.csv?$limit=1000000"

h = pd.read_csv(url, low_memory=False)

print("Download complete!")
print("Rows:", h.shape[0])
print("Columns:", h.shape[1])
```

```python

no_work_count = (
    h["job_type"].astype(str).str.strip().str.lower()
    == "no work"
).sum()

print("No Work records:", no_work_count)
```

```python

alt_co_new_building_count = (
    h["job_type"].astype(str).str.strip()
    == "ALT-CO - New Building with Existing Elements to Remain"
).sum()

print("ALT-CO New Building records:",
      alt_co_new_building_count)
```

```python
# Convert filing_date to proper datetime format
h["filing_date"] = pd.to_datetime(
    h["filing_date"],
    errors="coerce"
)

# Create job base and suffix
h["job_base"] = (
    h["job_filing_number"]
    .astype("string")
    .str.split("-")
    .str[0]
)

h["job_suffix"] = (
    h["job_filing_number"]
    .astype("string")
    .str.extract(r"-(.*)$")[0]
)

# Analyze repeated job bases
suffix_summary = h.groupby("job_base").agg(
    total_records=("job_filing_number", "size"),
    unique_suffixes=("job_suffix", "nunique"),
    earliest_filing=("filing_date", "min")
)

multiple_suffixes = suffix_summary[
    suffix_summary["unique_suffixes"] > 1
]

print("Unique job bases:", h["job_base"].nunique())
print("Job bases with multiple suffixes:", len(multiple_suffixes))

print("\nSuffix distribution:")
print(h["job_suffix"].value_counts().head(20))

display(multiple_suffixes.head(10))
```

```python
# Option A: Keep only initial I1 filings
option_a = h[
    h["job_filing_number"]
    .astype("string")
    .str.endswith("-I1", na=False)
].copy()

# Option B: Keep earliest filing per job_base
option_b = (
    h.dropna(subset=["filing_date"])
    .sort_values("filing_date")
    .drop_duplicates(subset="job_base", keep="first")
)

print("Original rows:", len(h))
print("Option A (-I1 only):", len(option_a))
print("Option B (earliest filing per base):", len(option_b))
```

```python
shed1 = h["shed"].astype("string").str.strip()
shed2 = h["sidewalk_shed_work_type_"].astype("string").str.strip()

both_present = shed1.notna() & shed2.notna()

same_shed = shed1[both_present] == shed2[both_present]

print("Both columns populated:", both_present.sum())
print("Identical among populated rows:",
      round(same_shed.mean() * 100, 2), "%")

print("\nOriginal shed column:")
print(h["shed"].value_counts(dropna=False).head(10))

print("\nSidewalk shed column:")
print(h["sidewalk_shed_work_type_"].value_counts(dropna=False).head(10))
```

```python
# 1. Start with a copy of the original data
housing_clean = h.copy()

# 2. Remove records with missing filing dates
missing_filing_dates = housing_clean["filing_date"].isna().sum()

housing_clean = housing_clean.dropna(
    subset=["filing_date"]
).copy()

print("Rows removed due to missing filing dates:",
      missing_filing_dates)

# 3. Deduplicate: earliest filing per job_base
housing_clean = (
    housing_clean
    .sort_values("filing_date")
    .drop_duplicates(subset="job_base", keep="first")
    .copy()
)

print("Rows after deduplication:", len(housing_clean))
print("Unique job bases:", housing_clean["job_base"].nunique())

# 4. Drop redundant shed column
housing_clean = housing_clean.drop(
    columns=["shed"]
)

# 5. Convert key numeric columns
numeric_columns = [
    "latitude",
    "longitude",
    "proposed_dwelling_units",
    "proposed_no_of_stories",
    "initial_cost"
]

for col in numeric_columns:
    housing_clean[col] = pd.to_numeric(
        housing_clean[col],
        errors="coerce"
    )

# 6. Set unrealistic dwelling units to missing
housing_clean.loc[
    (housing_clean["proposed_dwelling_units"] >= 998) &
    (housing_clean["proposed_height"] == 9999999),
    "proposed_dwelling_units"
] = np.nan

print("Final cleaned dataset shape:", housing_clean.shape)
```

```python
# Columns to check for outliers
outlier_columns = [
    "proposed_dwelling_units",
    "proposed_no_of_stories",
    "initial_cost"
]

for col in outlier_columns:

    Q1 = housing_clean[col].quantile(0.25)
    Q3 = housing_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    housing_clean[col + "_outlier"] = (
        (housing_clean[col] < lower_bound) |
        (housing_clean[col] > upper_bound)
    )

    print(f"\nColumn: {col}")
    print("Lower bound:", lower_bound)
    print("Upper bound:", upper_bound)
    print("Outliers:", housing_clean[col + "_outlier"].sum())
```

```python
# Check negative values in numeric columns
for col in [
    "proposed_dwelling_units",
    "proposed_no_of_stories",
    "initial_cost"
]:
    negative_count = (housing_clean[col] < 0).sum()

    print(f"{col}: {negative_count} negative values")
```

```python
# Check maximum values
for col in [
    "proposed_dwelling_units",
    "proposed_no_of_stories",
    "initial_cost",
    "proposed_height"
]:
    print(
        col,
        "Maximum:",
        housing_clean[col].max()
    )
```

```python
# Flag coordinates outside NYC's approximate bounding box
invalid_coordinates = (
    (housing_clean["latitude"] < 40.4774) |
    (housing_clean["latitude"] > 40.9176) |
    (housing_clean["longitude"] < -74.2591) |
    (housing_clean["longitude"] > -73.7004)
)

print("Coordinates outside NYC bounding box:",
      invalid_coordinates.sum())

print("Missing latitude:",
      housing_clean["latitude"].isna().sum())

print("Missing longitude:",
      housing_clean["longitude"].isna().sum())
```

```python
import geopandas as gpd

zip_gdf = gpd.read_file(
    "ZIP_Code_Tabulation_Areas_20261105.geojson"
)

print(zip_gdf.head())
print("\nColumns:", zip_gdf.columns.tolist())
print("\nCRS:", zip_gdf.crs)
```

```python
# Keep only the ZIP code and geometry
zip_gdf = zip_gdf[["zcta5", "geometry"]].copy()

# Ensure ZIP codes are 5-character strings
zip_gdf["zcta5"] = zip_gdf["zcta5"].astype(str).str.zfill(5)

print("ZIP boundary records:", len(zip_gdf))
print("CRS:", zip_gdf.crs)
```

```python
import pandas as pd
import geopandas as gpd

# Convert housing dataframe into a GeoDataFrame
housing_geo = gpd.GeoDataFrame(
    housing_clean.copy(),
    geometry=gpd.points_from_xy(
        housing_clean["longitude"],
        housing_clean["latitude"]
    ),
    crs="EPSG:4326"
)

# Keep only records with valid coordinates
housing_geo_valid = housing_geo[
    housing_geo.geometry.notna()
].copy()

print("Housing records with valid coordinates:",
      len(housing_geo_valid))
```

```python
# Match housing points to ZIP polygon boundaries
housing_geo_valid = gpd.sjoin(
    housing_geo_valid,
    zip_gdf,
    how="left",
    predicate="within"
)

print("Matched ZIP codes:",
      housing_geo_valid["zcta5"].notna().sum())

print("Unmatched coordinate records:",
      housing_geo_valid["zcta5"].isna().sum())
```

```python
# Make postcode a 5-digit string
housing_clean["postcode"] = (
    housing_clean["postcode"]
    .astype("Int64")
    .astype("string")
    .str.zfill(5)
)

# Map spatially matched ZIPs back to housing records
matched_zip = housing_geo_valid["zcta5"]

missing_zip_mask = housing_clean["postcode"].isna()

housing_clean.loc[
    missing_zip_mask,
    "postcode"
] = matched_zip.reindex(
    housing_clean.index
)[missing_zip_mask]

print("Missing ZIPs after filling:",
      housing_clean["postcode"].isna().sum())

print("Final ZIP column type:",
      housing_clean["postcode"].dtype)
```

```python
# Standardize borough names
housing_clean["borough"] = (
    housing_clean["borough"]
    .astype("string")
    .str.strip()
    .str.title()
)

# Extract year and month from filing_date
housing_clean["year"] = (
    housing_clean["filing_date"].dt.year.astype("Int64")
)

housing_clean["month"] = (
    housing_clean["filing_date"].dt.month.astype("Int64")
)

housing_clean["year_month"] = (
    housing_clean["filing_date"].dt.to_period("M").astype("string")
)

# Check results
print("Borough distribution:")
print(housing_clean["borough"].value_counts(dropna=False))

print("\nYear distribution:")
print(housing_clean["year"].value_counts().sort_index())

print("\nYear-month examples:")
print(housing_clean["year_month"].head())
```

```python
# Ensure date columns are datetime
date_columns = ["first_permit_date", "signoff_date"]

for col in date_columns:
    housing_clean[col] = pd.to_datetime(
        housing_clean[col], errors="coerce"
    )

# Calculate duration in days
housing_clean["days_permit_to_signoff"] = (
    housing_clean["signoff_date"] -
    housing_clean["first_permit_date"]
).dt.days

# Negative durations are invalid; replace with missing
housing_clean.loc[
    housing_clean["days_permit_to_signoff"] < 0,
    "days_permit_to_signoff"
] = np.nan

print("Missing permit-to-signoff durations:",
      housing_clean["days_permit_to_signoff"].isna().sum())

print(housing_clean["days_permit_to_signoff"].describe())
```

```python
# Select New Building and Full Demolition jobs
site_types = ["New Building", "Full Demolition"]

housing_sites_points = housing_clean[
    housing_clean["job_type"].isin(site_types) &
    housing_clean["latitude"].notna() &
    housing_clean["longitude"].notna()
].copy()

print("Housing site points:", len(housing_sites_points))

print("\nJob type distribution:")
print(housing_sites_points["job_type"].value_counts())

print("\nMissing coordinates:",
      housing_sites_points["latitude"].isna().sum())
```

```python
# Export the main cleaned dataset
housing_clean.to_csv(
    "housing_clean.csv",
    index=False
)

# Export the smaller map-ready dataset
housing_sites_points.to_csv(
    "housing_sites_points.csv",
    index=False
)

print("Both CSV files exported successfully!")
```
